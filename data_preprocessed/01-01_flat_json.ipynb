```python
import shutil
from pathlib import Path

# 원본 JSON이 들어있는 루트 폴더
SRC_ROOT = Path(r"C:/project_sep/flat_json")

# 평탄화해서 저장할 폴더
DST_JSON = Path(r"C:/project_sep/data/flat_json")
DST_JSON.mkdir(parents=True, exist_ok=True)

count, skipped = 0, 0

# VL_*_단일 하위의 모든 json 파일 탐색
for jpath in SRC_ROOT.rglob("*.json"):
    try:
        # 파일명만 그대로 복사 (폴더 경로 제거)
        shutil.copy(jpath, DST_JSON / jpath.name)
        count += 1
    except shutil.SameFileError:
        skipped += 1
    except Exception as e:
        print(f"[스킵] {jpath} → {e}")
        skipped += 1

print("=== JSON Flatten 완료 ===")
print("총 복사된 JSON 수:", count)
print("스킵된 파일 수:", skipped)
print("저장 위치:", DST_JSON)
```

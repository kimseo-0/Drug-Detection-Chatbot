```python
painkiller_general_white['folder_prefix'] = painkiller_general_white['folder'].str.split('_').str[0]
painkiller_general_white['folder_num'] = painkiller_general_white['folder'].str.split('_').str[1].astype(int)
```

```python
code_map = {}
for key in folder_names:
    code_map[key] = painkiller_general_white.loc[painkiller_general_white['folder'] == key, 'c-code'].tolist()
```

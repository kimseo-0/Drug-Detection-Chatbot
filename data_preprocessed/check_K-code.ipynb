```python
import os
import re
from collections import defaultdict

# single_folder 경로
json_dir = "C:/project_sep/data/flat_json"

# 사용할 K-code 리스트
target_kcodes = [
    "K-003361",
    "K-004578",
    "K-009429",
    "K-010281",
    "K-011274",
    "K-011566",
    "K-011833",
    "K-015270",
    "K-022362",
    "K-000105",
    "K-000271",
    "K-000304",
    "K-000341",
    "K-000608",
    "K-000653",
    "K-000937",
    "K-001516",
    "K-001670",
    "K-002896",
    "K-002937",
    "K-003244",
    "K-004072",
    "K-004469",
    "K-006677",
    "K-011701",
    "K-014206",
    "K-014383",
    "K-015197",
    "K-021984",
    "K-022244",
    "K-023679",
    "K-030912",
    "K-035610",
    "K-035811",
    "K-042098",
    "K-043149",
    "K-044927",
]

# 개수 저장
kcode_counts = defaultdict(int)

# 정규식: 파일명 맨 앞에서 'K-숫자6자리' 추출
pattern = re.compile(r"^(K-\d{6})")

for fname in os.listdir(json_dir):
    match = pattern.match(fname)
    if match:
        kcode = match.group(1)
        if kcode in target_kcodes:
            kcode_counts[kcode] += 1

# 결과 출력
for k in target_kcodes:
    print(f"{k}: {kcode_counts[k]} 개")

print("\n총 이미지 개수:", sum(kcode_counts.values()))
```

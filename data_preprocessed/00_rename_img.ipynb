# 폴더 안에 이미지/json 이름 변경

## 바뀔 파일 이름 미리 확인하기

```python
import os
import re

# 작업할 폴더 경로
folder = "C:/project_sep/data/_flat_images"

# 변환 결과 미리보기
for filename in os.listdir(folder):
    if filename.lower().endswith((".png", ".jpg", ".jpeg")):
        # 중복된 K-code 제거
        new_name = re.sub(r"^(K-\d+(?:-\d+)*?)_\1", r"\1", filename)

        if filename != new_name:
            print(f"[변경 예정] {filename}  →  {new_name}")
        else:
            print(f"[변경 없음] {filename}")
```

## 실제로 변경

```python
import os
import re

# 작업할 폴더 경로
folder = "C:/project_sep/data/_flat_images"

# 파일 순회
for filename in os.listdir(folder):
    old_path = os.path.join(folder, filename)

    # 확장자가 png/jpg/jpeg 인 경우만 처리
    if os.path.isfile(old_path) and filename.lower().endswith(
        (".png", ".jpg", ".jpeg")
    ):
        # 중복된 K-code 제거 (예: K-xxxx_K-xxxx → K-xxxx)
        new_name = re.sub(r"^(K-\d+(?:-\d+)*?)_\1", r"\1", filename)

        new_path = os.path.join(folder, new_name)

        # 이름이 달라진 경우만 rename
        if old_path != new_path:
            os.rename(old_path, new_path)
            print(f"Renamed: {filename}  →  {new_name}")
```

## 매칭이 잘 되었는지 확인하기

```python
import os

img_folder = "C:/project_sep/data/_flat_images"
json_folder = "C:/project_sep/data/_flat_json"

# 이미지 파일 이름 (확장자 제거)
img_names = {
    os.path.splitext(f)[0]
    for f in os.listdir(img_folder)
    if f.lower().endswith((".png", ".jpg", ".jpeg"))
}

# JSON 파일 이름 (확장자 제거)
json_names = {
    os.path.splitext(f)[0]
    for f in os.listdir(json_folder)
    if f.lower().endswith(".json")
}

# 매칭 확인
missing_json = img_names - json_names
missing_img = json_names - img_names

print(f"이미지 개수: {len(img_names)}")
print(f"JSON 개수: {len(json_names)}\n")

if missing_json:
    print("JSON 폴더에 없는 이미지들:")
    for name in sorted(missing_json)[:20]:
        print(" ", name)
else:
    print("모든 이미지에 JSON 있음")

if missing_img:
    print("\n이미지 폴더에 없는 JSON들:")
    for name in sorted(missing_img)[:20]:
        print(" ", name)
else:
    print("\n모든 JSON에 이미지 있음")
```

```python
import os, re

img_folder = "C:/project_sep/data/_flat_images"
json_folder = "C:/project_sep/data/_flat_json"

# 이미지 파일 이름 (확장자 제거)
img_names = {
    os.path.splitext(f)[0]
    for f in os.listdir(img_folder)
    if f.lower().endswith((".png", ".jpg", ".jpeg"))
}

# JSON 파일 이름 (확장자 제거)
json_names = {
    os.path.splitext(f)[0]
    for f in os.listdir(json_folder)
    if f.lower().endswith(".json")
}

# 이미지에는 없고 JSON에만 있는 이름들
missing_img = json_names - img_names

for filename in os.listdir(json_folder):
    if filename.lower().endswith(".json"):
        name_wo_ext = os.path.splitext(filename)[0]
        if name_wo_ext in missing_img:
            old_path = os.path.join(json_folder, filename)
            # 맨 앞 K-숫자_ 제거
            new_name = re.sub(r"^K-\d+_", "", filename)
            new_path = os.path.join(json_folder, new_name)

            if filename != new_name:
                os.rename(old_path, new_path)
                print(f"Renamed: {filename} → {new_name}")
```

```python
conflicts = []

if filename != new_name:
    if os.path.exists(new_path):
        conflicts.append((filename, new_name))
    else:
        os.rename(old_path, new_path)
        print(f"Renamed: {filename} → {new_name}")

if conflicts:
    print("\n충돌된 파일들:")
    for old, new in conflicts:
        print(f" {old} → {new} (이미 존재)")
```

## 충돌된 이미지/json 파일 삭제

```python
import os, re

json_folder = "C:/project_sep/data/_flat_json"

for filename in os.listdir(json_folder):
    old_path = os.path.join(json_folder, filename)

    if os.path.isfile(old_path) and filename.lower().endswith(".json"):
        # 맨 앞 K-숫자_ 제거
        new_name = re.sub(r"^K-\d+_", "", filename)
        new_path = os.path.join(json_folder, new_name)

        if filename != new_name:
            # 이미 존재한다면 → 기존 파일 삭제
            if os.path.exists(new_path):
                os.remove(new_path)
                print(f"🗑 기존 파일 삭제: {new_name}")

            # rename 진행
            os.rename(old_path, new_path)
            print(f"Renamed: {filename} → {new_name}")
```

```python
import os, re

json_folder = "C:/project_sep/data/_flat_json"

for filename in os.listdir(json_folder):
    old_path = os.path.join(json_folder, filename)

    if os.path.isfile(old_path) and filename.lower().endswith(".json"):
        # 1) 맨 앞 K-숫자_ 제거
        new_name = re.sub(r"^K-\d+_", "", filename)
        # 2) 맨 앞 json_ 제거
        new_name = re.sub(r"^json_", "", new_name)

        new_path = os.path.join(json_folder, new_name)

        if filename != new_name:
            if os.path.exists(new_path):
                os.remove(new_path)
                print(f"🗑 기존 파일 삭제: {new_name}")

            os.rename(old_path, new_path)
            print(f"Renamed: {filename} → {new_name}")
```

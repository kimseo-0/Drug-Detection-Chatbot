```python
import os
import shutil

img_folder = "C:/project_sep/data/_flat_images"
single_folder = "C:/project_sep/data/single_folder"
combo_folder = "C:/project_sep/data/combo_folder"

# 결과 저장 폴더 생성
os.makedirs(single_folder, exist_ok=True)
os.makedirs(combo_folder, exist_ok=True)

for filename in os.listdir(img_folder):
    if filename.lower().endswith((".png")):
        # 확장자 제외한 이름
        name_wo_ext = os.path.splitext(filename)[0]

        # "K-" 뒤의 숫자들만 추출
        parts = name_wo_ext.split("_")[0]
        k_codes = parts.split("-")[1:]

        old_path = os.path.join(img_folder, filename)

        if len(k_codes) == 1:
            # 단일
            new_path = os.path.join(single_folder, filename)
            shutil.move(old_path, new_path)
            print(f"단일 -> {filename}")
        else:
            # 조합
            new_path = os.path.join(combo_folder, filename)
            shutil.move(old_path, new_path)
            print(f"조합 -> {filename}")
```

# 특정 폴더 내 모든 폴더 리스트

```python
import os

def list_subfolders(parent_dir: str) -> list[str]:
    return [
        name for name in os.listdir(parent_dir)
        if os.path.isdir(os.path.join(parent_dir, name))
    ]
```

# 특정 폴더 특정 확장 파일 리스트

```python
import os
from typing import List

def list_files_with_ext(folder_path: str, extension: str) -> List[str]:
    extension = extension.lower()
    return [
        os.path.join(folder_path, f)
        for f in os.listdir(folder_path)
        if f.lower().endswith(extension)
    ]
```

```python
import json
import csv

def get_json_to_dict(json_path: str, csv_path: str):
    # JSON 불러오기
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    img = data['images'][0]
    type = data['type']
    ann = data["annotations"][0]
    category = data["categories"][0]

    return {
        "id": img.get("id"),

        "type": type,

        "file_name": img.get("file_name"),
        "width": img.get("width"),
        "height": img.get("height"),
        "imgfile": img.get("imgfile"),
        "drug_N": img.get("drug_N"),
        "drug_S": img.get("drug_S"),
        "back_color": img.get("back_color"),
        "drug_dir": img.get("drug_dir"),
        "light_color": img.get("light_color"),
        "camera_la": img.get("camera_la"),
        "camera_lo": img.get("camera_lo"),
        "size": img.get("size"),
        "item_seq": img.get("item_seq"),
        "leng_long": img.get("leng_long"),
        "leng_short": img.get("leng_short"),
        "thick": img.get("thick"),
        "dl_mapping_code": img.get("dl_mapping_code"),
        "dl_idx": img.get("dl_idx"),
        "dl_name": img.get("dl_name"),
        "dl_name_en": img.get("dl_name_en"),
        "dl_material": img.get("dl_material"),
        "dl_material_en": img.get("dl_material_en"),
        "dl_custom_shape": img.get("dl_custom_shape"),
        "dl_company": img.get("dl_company"),
        "dl_company_en": img.get("dl_company_en"),
        "di_company_mf": img.get("di_company_mf"),
        "di_company_mf_en": img.get("di_company_mf_en"),
        "di_item_permit_date": img.get("di_item_permit_date"),
        "di_class_no": img.get("di_class_no"),
        "di_etc_otc_code": img.get("di_etc_otc_code"),
        "di_edi_code": img.get("di_edi_code"),
        "chart": img.get("chart"),
        "drug_shape": img.get("drug_shape"),
        "print_front": img.get("print_front"),
        "print_back": img.get("print_back"),
        "color_class1": img.get("color_class1"),
        "color_class2": img.get("color_class2"),
        "line_front": img.get("line_front"),
        "line_back": img.get("line_back"),
        "img_regist_ts": img.get("img_regist_ts"),
        "form_code_name": img.get("form_code_name"),
        "mark_code_front_anal": img.get("mark_code_front_anal"),
        "mark_code_back_anal": img.get("mark_code_back_anal"),
        "mark_code_front_img": img.get("mark_code_front_img"),
        "mark_code_back_img": img.get("mark_code_back_img"),
        "mark_code_front": img.get("mark_code_front"),
        "mark_code_back": img.get("mark_code_back"),
        "change_date": img.get("change_date"),

        "area": ann.get("area"),
        "iscrowd": ann.get("iscrowd"),
        "bbox": ann.get("bbox"),
        "category_id": ann.get("category_id"),
        "ignore": ann.get("ignore"),
        "segmentation": ann.get("segmentation"),
        "image_id": ann.get("image_id"),

        "supercategory": category.get("supercategory"),
        "category_name": category.get("name"),
    }

def append_dict_to_csv(row, csv_path):
    # CSV 헤더 정의
    fieldnames = row.keys()

    # 파일 있으면 append, 없으면 새로 생성
    file_exists = os.path.isfile(csv_path)
    with open(csv_path, "a", newline="", encoding="utf-8") as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

        if not file_exists:  # 처음이면 헤더 추가
            writer.writeheader()
            writer.writerow(row.keys())
        
        writer.writerows([row])

    print(f"CSV에 행 추가 완료 → {csv_path}")
```

# 실행

```python
base_dir = r"C:\\Potenup\Drug-Detection-Chatbot\data\\json_unzip"
```

```python
folders = list_subfolders(base_dir)
print(folders)
```

```python
files = {}
for folder_name in folders:
    folder_path = os.path.join(base_dir, folder_name)
    file_names = list_subfolders(folder_path)
    files[folder_name] = file_names
```

```python
csv_path = "C:\Potenup\Drug-Detection-Chatbot\data\\drug_json_data.csv"
```

```python
for key, value in files.items():
    print(key)
    for i, fold_name in enumerate(value):
        folder_path = os.path.join(base_dir, key, fold_name)
        json_files = list_files_with_ext(folder_path, ".json")
        
        json_file_path = os.path.join(folder_path, json_files[0])
        print(f'{key} {fold_name} {i + 1}/{len(value)}', end=' ')
        row = get_json_to_dict(json_file_path, csv_path)
        row['folder'] = key
        row['c-code'] = fold_name.split('_')[0]
        append_dict_to_csv(row, csv_path)
```
